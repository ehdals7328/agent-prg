```python
import os
from dotenv import load_dotenv

# .env 파일 로드
load_dotenv()

# 환경 변수 값 가져오기
api_key = os.getenv("OPENAI_API_KEY")

if api_key:
    # 보안을 위해 전체를 출력하지 않고 앞 8자리만 출력
    print(f"✅ OPENAI_API_KEY 인식 성공: {api_key[:8]}...")
else:
    print("❌ OPENAI_API_KEY를 찾을 수 없습니다. .env 파일 경로나 변수명을 확인하세요.")
```
